## Project Summary

This proposal outlines a code-first AI agent that answers common support and platform-policy questions for Sync In Social users. The agent will retrieve approved internal support documentation, classify issue types, explain relevant policy or feature rules, and recommend escalation to a human when needed.

**Company Name:** Sync In Social

**Industry:** Social media, creator networking, and online community platforms

Sync In Social is a social media and community networking platform that supports user profiles, posts, comments, replies, direct messages, groups, clips, business pages, content moderation, and notifications. Because this is our own application, the team can use it as a realistic business case and, if appropriate, demonstrate the agent with a live or production-style version of the platform.

### 2. Use Case and Business Value

***Use Case: Customer Support and Platform Policy Assistant***

The agent would help users and internal support staff answer common questions such as:
* "Why can't I upload a post?"
* "Why is my content under review?"
* "Why can't I message someone?"
* "How do I change my profile settings?"
* "Why can't I access a group feature?"

These questions are important because a social platform can quickly receive repetitive support requests as it grows.

This use case is especially relevant for Sync In Social because the platform has custom rules around profile completion, content moderation, post visibility, direct messages, groups, clips, media uploads, and business pages. A generic third-party chatbot would not understand these platform-specific rules unless it was given access to the correct documentation and policies. Building our own support agent allows the team to control the knowledge base, define escalation rules, protect sensitive information, and update the system as the platform evolves.

***Estimated ROI***

ROI can be estimated by comparing the cost of human-handled support tickets with the cost of AI-assisted responses. For example, if a human support ticket costs approximately $5 and an AI response costs only a few cents, deflecting 60% of 1,000 monthly tickets could save close to $3,000 per month before accounting for development and infrastructure costs. This estimate can be adjusted as the team gathers better assumptions about ticket volume, average handling cost, model cost, and escalation rate.

| ROI Input | Example Assumption |
|-----------|-------------------|
| Monthly support questions | 1,000 |
| Target AI resolution rate | 60% |
| Estimated human ticket cost | $5 per ticket |
| Estimated AI response cost | A few cents per response |
| Potential monthly savings | Approximately $3,000 before overhead |

### 3. Dataset

The dataset will be built from synthetic and internally created documents instead of private user data. This makes the project realistic while staying manageable for the course timeline and avoiding unnecessary privacy risk.

**Proposed Data Sources**

* Synthetic support tickets and sample user questions.
* Platform FAQ content and profile setup instructions.
* Content moderation and posting policy summaries.
* Direct message, group, clip, media upload, and business page rules.
* Sample escalation scenarios for issues that require human review.
* Approved short answers that can be used as evaluation examples.

**Why This Dataset Is Appropriate**

This dataset is appropriate because it directly matches the agent's support use case. The agent needs to understand platform features, policies, and troubleshooting steps, so a curated support and policy knowledge base is more useful than a generic public dataset. The dataset is also reasonable for the course because the team can create a focused collection of documents and test questions without needing production user data.

**Tooling for the Dataset**

* Databricks notebooks for preparing support documents and synthetic tickets.
* Delta tables for storing documents, ticket examples, labels, and evaluation results.
* Embeddings and vector search for retrieving relevant support or policy content.
* MLflow traces for logging agent steps, retrieved context, model outputs, cost, latency, and evaluation results.
* A simple evaluation set with expected answers, escalation labels, and rubric-based quality scores.

### 4. High-Level Technical Plan

The agent will use a retrieval-augmented, ReAct-style design. The LLM will analyze the user's question, decide which tool to use, retrieve relevant documentation, apply platform rules, and then generate a grounded answer or recommend escalation to a human support process.

**Intended Tools**

* **Issue Type Classifier:** identifies whether the question is about posting, moderation, profile setup, messaging, groups, clips, business pages, or account access.
* **Document Search Tool:** retrieves relevant support or policy documents using vector search.
* **Policy Lookup Tool:** returns the most relevant platform rule or troubleshooting step.
* **Escalation Recommendation Tool:** flags questions that involve account-specific issues, safety concerns, appeals, payment disputes, or unclear policy cases.
* **Evaluation and Trace Logging:** stores the agent's reasoning path, tool calls, retrieved context, final answer, and quality scores.

**LLM Plan**

The team will test two specific LLM options to compare answer quality, cost, latency, grounding, and escalation accuracy. The primary quality model will be **GPT-4.1 mini** because it is strong enough for policy-sensitive support answers while still being reasonable for a course prototype. The cost-efficient comparison model will be **Meta Llama 3.1 70B Instruct**, or the closest Databricks-hosted open model available in the class workspace. If those exact endpoints are unavailable, the team will keep the same experiment design by selecting the strongest available model and a lower-cost available model, then comparing results using the same evaluation set.

**Evaluation Plan**

* **Answer correctness:** whether the response matches the policy or support documentation.
* **Grounding quality:** whether the response is based on retrieved documentation rather than unsupported assumptions.
* **Escalation accuracy:** whether the agent correctly routes account-specific or sensitive questions to human support.
* **Helpfulness and clarity:** whether the answer is understandable to a normal user.
* **Safety and policy compliance:** whether the agent avoids giving incorrect, private, or harmful guidance.
* **Operational metrics:** cost per response, latency, retrieval quality, and failure rate.

### 6. Agentic Design Pattern

The intended design pattern is a **ReAct-style support agent** combined with **retrieval-augmented generation**. ReAct fits this use case because the agent needs to reason about the question, choose tools, retrieve platform documentation, and decide whether it can answer directly or should escalate the case. This pattern is more reliable than a simple chatbot because the agent's answer can be grounded in approved support and policy content.

The design also supports a practical production workflow. Each response can be logged through MLflow so the team can inspect what the agent retrieved, what tool it used, how it answered, and whether the answer met the evaluation rubric. This makes the agent easier to improve because failures can be traced back to missing documentation, weak retrieval, unclear policies, or model mistakes.
